```python
import os
import numpy as np
import tensorflow as tf
import librosa
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
from sklearn.preprocessing import StandardScaler
# --- MODIFIED: Added classification_report ---
from sklearn.metrics import confusion_matrix, f1_score, classification_report
from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras.layers import (Input, Dense, BatchNormalization, LeakyReLU,
                                     Conv1D, MultiHeadAttention, Multiply, Add,
                                     Reshape, Flatten, Layer)
from tensorflow.keras.models import Model
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping
import tensorflow.keras.backend as K

# Ensure reproducibility
tf.random.set_seed(42)
np.random.seed(42)

# --- 1. Feature Extraction and Data Loading Functions ---
def extract_features(file_path, max_len=130):
    y, sr = librosa.load(file_path, sr=None)
    mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=20)
    mfccs_delta = librosa.feature.delta(mfccs)
    mfccs_delta2 = librosa.feature.delta(mfccs, order=2)
    zcr = librosa.feature.zero_crossing_rate(y=y).flatten()
    chroma = librosa.feature.chroma_stft(y=y, sr=sr)
    spectral_contrast = librosa.feature.spectral_contrast(y=y, sr=sr)
    spectral_centroid = librosa.feature.spectral_centroid(y=y, sr=sr)
    spectral_rolloff = librosa.feature.spectral_rolloff(y=y, sr=sr)
    spectral_bandwidth = librosa.feature.spectral_bandwidth(y=y, sr=sr)
    tonnetz = librosa.feature.tonnetz(y=y, sr=sr)
    features_list = []
    def pad_or_truncate(feature_array, target_len, axis=1):
        if feature_array.shape[axis] < target_len:
            pad_width = target_len - feature_array.shape[axis]
            if axis == 1:
                feature_array = np.pad(feature_array, pad_width=((0, 0), (0, pad_width)), mode='constant')
            else:
                feature_array = np.pad(feature_array, pad_width=(0, pad_width), mode='constant')
        else:
            if axis == 1:
                feature_array = feature_array[:, :target_len]
            else:
                feature_array = feature_array[:target_len]
        return feature_array
    features_list.append(pad_or_truncate(mfccs, max_len))
    features_list.append(pad_or_truncate(mfccs_delta, max_len))
    features_list.append(pad_or_truncate(mfccs_delta2, max_len))
    features_list.append(pad_or_truncate(zcr[np.newaxis, :], max_len))
    features_list.append(pad_or_truncate(chroma, max_len))
    features_list.append(pad_or_truncate(spectral_contrast, max_len))
    features_list.append(pad_or_truncate(spectral_centroid, max_len))
    features_list.append(pad_or_truncate(spectral_rolloff, max_len))
    features_list.append(pad_or_truncate(spectral_bandwidth, max_len))
    features_list.append(pad_or_truncate(tonnetz, max_len))
    combined_features = np.concatenate(features_list, axis=0).T
    return combined_features

def load_data(data_dir, max_len=130):
    features = []
    labels = []
    emotion_map = {'Angry': 0, 'Happy': 1, 'Anxiety': 2, 'Sad': 3, 'Neutral': 4}
    inverse_emotion_map = {v: k for k, v in emotion_map.items()}
    actual_num_features = None
    sample_file_path = None
    for label_name in os.listdir(data_dir):
        label_path = os.path.join(data_dir, label_name)
        if os.path.isdir(label_path):
            for filename in os.listdir(label_path):
                if filename.endswith('.wav'):
                    sample_file_path = os.path.join(label_path, filename)
                    break
        if sample_file_path:
            break
    if sample_file_path:
        try:
            temp_feature = extract_features(sample_file_path, max_len=max_len)
            actual_num_features = temp_feature.shape[1]
            print(f"Actual number of features determined: {actual_num_features}")
        except Exception as e:
            print(f"Could not determine feature count from sample file {sample_file_path}: {e}")
            print("Defaulting to 89. This may cause errors if feature count differs.")
            actual_num_features = 89
    else:
        print("No sample WAV file found to determine feature count. Defaulting to 89.")
        actual_num_features = 89
    expected_shape = (max_len, actual_num_features)
    for label_name in tqdm(os.listdir(data_dir), desc=f"Loading data from {os.path.basename(data_dir)}"):
        label_path = os.path.join(data_dir, label_name)
        if os.path.isdir(label_path):
            for filename in os.listdir(label_path):
                if filename.endswith('.wav'):
                    file_path = os.path.join(label_path, filename)
                    try:
                        feature = extract_features(file_path, max_len=max_len)
                        if feature.shape == expected_shape:
                            features.append(feature)
                            labels.append(emotion_map[label_name])
                        else:
                            print(f"Skipping {file_path} due to incorrect feature shape: {feature.shape}. Expected: {expected_shape}")
                    except Exception as e:
                        print(f"Error processing {file_path}: {e}")
    return np.array(features), np.array(labels), inverse_emotion_map, actual_num_features


# --- 2. STACN Custom Layers  ---
class SparseDropout(Layer):
    def __init__(self, rate=0.8, **kwargs):
        super(SparseDropout, self).__init__(**kwargs)
        self.rate = rate
    def call(self, inputs, training=None):
        if training:
            keep_prob = 1 - self.rate
            random_tensor = tf.random.uniform(shape=tf.shape(inputs))
            binary_mask_bool = random_tensor < keep_prob
            binary_mask = tf.cast(binary_mask_bool, dtype=inputs.dtype)
            return inputs * binary_mask
        return inputs
    def get_config(self):
        config = super(SparseDropout, self).get_config()
        config.update({'rate': self.rate})
        return config

class CapsuleLayer(Layer):
    def __init__(self, num_output_capsules, capsule_dim, num_routing_iterations=3, **kwargs):
        super(CapsuleLayer, self).__init__(**kwargs)
        self.num_output_capsules = num_output_capsules
        self.capsule_dim = capsule_dim
        self.num_routing_iterations = num_routing_iterations
    def build(self, input_shape):
        self.input_capsule_dim = input_shape[-1]
        self.num_input_capsules = input_shape[1]
        self.kernel = self.add_weight(shape=(self.num_input_capsules,
                                             self.num_output_capsules,
                                             self.input_capsule_dim,
                                             self.capsule_dim),
                                      initializer='glorot_uniform',
                                      name='kernel')
        self.built = True
    def call(self, inputs):
        votes = tf.einsum('bid,ijdo->bijo', inputs, self.kernel)
        b_ij = tf.zeros(shape=[tf.shape(inputs)[0], self.num_input_capsules, self.num_output_capsules])
        for r_iter in range(self.num_routing_iterations):
            c_ij = tf.nn.softmax(b_ij, axis=2)
            c_ij_expand = tf.expand_dims(c_ij, axis=-1)
            s_j = tf.reduce_sum(c_ij_expand * votes, axis=1)
            norm_s_j = tf.sqrt(tf.reduce_sum(tf.square(s_j), axis=-1, keepdims=True) + K.epsilon())
            v_j = (norm_s_j**2 / (1 + norm_s_j**2)) * (s_j / norm_s_j)
            if r_iter < self.num_routing_iterations - 1:
                v_j_expand = tf.expand_dims(v_j, axis=1)
                agreement = tf.reduce_sum(votes * v_j_expand, axis=-1)
                b_ij = b_ij + agreement
        return v_j
    def compute_output_shape(self, input_shape):
        return (input_shape[0], self.num_output_capsules, self.capsule_dim)
    def get_config(self):
        config = super(CapsuleLayer, self).get_config()
        config.update({
            'num_output_capsules': self.num_output_capsules,
            'capsule_dim': self.capsule_dim,
            'num_routing_iterations': self.num_routing_iterations,
        })
        return config

# --- 3. STACN Model Definition ---
def build_stacn(input_shape, num_classes, config):
    """
    Builds the Sparse Temporal-Aware Capsule Network (STACN) model.
    input_shape: (time_steps, num_features)
    config: Dictionary containing model hyperparameters.
    """
    inputs = tf.keras.Input(shape=input_shape)

    # 1. Sparsity-Enhanced Feature Mapping Module
    x = Dense(128)(inputs)
    x = BatchNormalization()(x)
    x = LeakyReLU(negative_slope=0.2)(x)
    h_s = SparseDropout(rate=config['sparse_dropout_rate'])(x)

    # Repeat the dense block as in the diagram (x2)
    x = Dense(128)(h_s)
    x = BatchNormalization()(x)
    x = LeakyReLU(negative_slope=0.2)(x)
    h_s = SparseDropout(rate=config['sparse_dropout_rate'])(x)

    # 2. Multi-Head Attention (MHA) Mechanism
    h_m = MultiHeadAttention(num_heads=config['num_heads_mha'], key_dim=128 // config['num_heads_mha'])(h_s, h_s, h_s)

    # 3. Temporal-Aware Block (TAB)
    # First: Dilated 1D convolution with dilation_rate=1
    h_c = Conv1D(filters=128, kernel_size=3, padding='same', dilation_rate=1)(h_m)
    h_b = BatchNormalization()(h_c)
    h_a = LeakyReLU(negative_slope=0.2)(h_b)
    h_d1 = SparseDropout(rate=config['sparse_dropout_rate'])(h_a)

    # Second: Dilated 1D convolution with dilation_rate=2
    h_c2 = Conv1D(filters=128, kernel_size=3, padding='same', dilation_rate=2)(h_d1)
    h_b2 = BatchNormalization()(h_c2)
    h_a2 = LeakyReLU(negative_slope=0.2)(h_b2)
    h_d2 = SparseDropout(rate=config['sparse_dropout_rate'])(h_a2)

    # Apply Sigmoid and element-wise multiplication for the temporal-aware output
    S = tf.keras.activations.sigmoid(h_d2)
    h_m_prime = h_m
    h_t = Multiply()([h_m_prime, S])

    # 4. Residual module
    # Chain of Conv1D layers with decreasing filters
    h_r1 = Conv1D(filters=1024, kernel_size=3, padding='same')(h_t)
    h_r1 = LeakyReLU(negative_slope=0.2)(h_r1)
    h_r1 = BatchNormalization()(h_r1)

    h_r2 = Conv1D(filters=512, kernel_size=3, padding='same')(h_r1)
    h_r2 = LeakyReLU(negative_slope=0.2)(h_r2)
    h_r2 = BatchNormalization()(h_r2)

    h_r3 = Conv1D(filters=256, kernel_size=3, padding='same')(h_r2)
    h_r3 = LeakyReLU(negative_slope=0.2)(h_r3)
    h_r3 = BatchNormalization()(h_r3)

    # Adjustment layer (1x1 conv) for the residual connection
    h_a = Conv1D(filters=256, kernel_size=1, padding='same')(h_t)
    h_a = LeakyReLU(negative_slope=0.2)(h_a)

    # Residual connection
    h_r = Add()([h_r3, h_a])

    # 5. Capsule Module
    # Reshape h_r (None, time_steps, 256) into primary capsules
    primary_capsule_dimension = config['primary_capsule_dimension']
    num_primary_capsules_per_timestep = 256 // primary_capsule_dimension
    num_total_primary_capsules = input_shape[0] * num_primary_capsules_per_timestep

    primary_capsules = Reshape((num_total_primary_capsules, primary_capsule_dimension))(h_r)

    # Apply CapsuleLayer with Dynamic Routing
    output_capsules = CapsuleLayer(num_output_capsules=num_classes,
                                     capsule_dim=primary_capsule_dimension,
                                     num_routing_iterations=config['num_routing_iterations']
                                    )(primary_capsules)

    flattened_capsules = Flatten()(output_capsules)

    # 6. Output Layer
    outputs = Dense(num_classes, activation='softmax')(flattened_capsules)

    model = Model(inputs=inputs, outputs=outputs)
    return model


# --- Evaluation and Visualization Function ---
def evaluate_and_visualize_model(model, history, X_test, y_test, emotion_labels, config):
    """
    Evaluates the model on the test/validation set and visualizes the results.
    """
    print("\n--- Evaluating model on the Test/Validation set (using best weights) ---")

    batch_size = config['batch_size']
    loss, accuracy = model.evaluate(X_test, y_test, batch_size=batch_size, verbose=0)
    
    print(f"Test/Validation Loss: {loss:.4f}")
    print(f"Test/Validation Accuracy: {accuracy:.4f}")

    # Get predictions
    y_pred_probs = model.predict(X_test, batch_size=batch_size)
    y_pred = np.argmax(y_pred_probs, axis=1)
    y_true = np.argmax(y_test, axis=1)

    # Visualization
    # Loss Plot
    plt.figure(figsize=(12, 6))
    plt.plot(history.history['loss'], label='Training Loss')
    plt.plot(history.history['val_loss'], label='Validation Loss')
    plt.title('STACN Model Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    plt.savefig('stacn_loss_plot.png')
    plt.show()

    # Accuracy Plot
    plt.figure(figsize=(12, 6))
    plt.plot(history.history['accuracy'], label='Training Accuracy')
    plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
    plt.title('STACN Model Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True)
    plt.savefig('stacn_accuracy_plot.png')
    plt.show()

    # Confusion Matrix
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=list(emotion_labels.values()),
                yticklabels=list(emotion_labels.values()))
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    plt.title('Confusion Matrix for STACN Model')
    plt.savefig('stacn_confusion_matrix.png')
    plt.show()

    # --- NEW: Use classification_report for detailed metrics ---
    print("\nClassification Report:")
    target_names = list(emotion_labels.values())
    # Ensure the labels match the order of target_names
    report = classification_report(y_true, y_pred, target_names=target_names, labels=np.arange(len(target_names)))
    print(report)


# --- Main Execution Block ---
def main():
    CONFIG = {
        "train_data_path": '/kaggle/input/speech-emotion-10/VESC/VESC',
        "test_data_path": '/kaggle/input/speech-emotion-10/VESC_test/VESC_test',
        "max_len": 500,
        "epochs": 100,
        "batch_size": 16,
        "learning_rate": 0.0001,
        "sparse_dropout_rate": 0.5,
        "num_heads_mha": 8,
        "primary_capsule_dimension": 16,
        "num_routing_iterations": 3
    }

    print("Loading training data...")
    X_train_raw, y_train_raw, emotion_labels, actual_num_features = load_data(CONFIG['train_data_path'], max_len=CONFIG['max_len'])
    print("Loading test data...")
    X_test_raw, y_test_raw, _, _ = load_data(CONFIG['test_data_path'], max_len=CONFIG['max_len'])

    print("\n--- Training Data Distribution ---")
    unique, counts = np.unique(y_train_raw, return_counts=True)
    class_distribution = dict(zip(unique, counts))
    for emotion_code, count in class_distribution.items():
        emotion_name = emotion_labels[emotion_code]
        print(f"{emotion_name}: {count} samples")

    plt.figure(figsize=(10, 5))
    plt.bar([emotion_labels[i] for i in unique], counts)
    plt.title('Class Distribution in Training Set')
    plt.xlabel('Emotion')
    plt.ylabel('Number of Samples')
    plt.show()

    if X_train_raw.size == 0 or X_test_raw.size == 0:
        print("Not enough valid data found.")
        return

    scaler = StandardScaler()
    X_train_reshaped = X_train_raw.reshape(-1, X_train_raw.shape[-1])
    X_test_reshaped = X_test_raw.reshape(-1, X_test_raw.shape[-1])
    X_train = scaler.fit_transform(X_train_reshaped).reshape(X_train_raw.shape)
    X_test = scaler.transform(X_test_reshaped).reshape(X_test_raw.shape)

    num_classes = len(np.unique(y_train_raw))
    y_train = tf.keras.utils.to_categorical(y_train_raw, num_classes=num_classes)
    y_test = tf.keras.utils.to_categorical(y_test_raw, num_classes=num_classes)

    input_shape = (X_train.shape[1], X_train.shape[2])
    stacn_model = build_stacn(input_shape, num_classes, CONFIG)
    stacn_model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=CONFIG['learning_rate']),
                          loss='categorical_crossentropy',
                          metrics=['accuracy'])
    stacn_model.summary()

    class_weights = compute_class_weight(
        class_weight='balanced',
        classes=np.unique(y_train_raw),
        y=y_train_raw
    )
    class_weight_dict = dict(enumerate(class_weights))
    print(f"\nCalculated Class Weights to counter imbalance: {class_weight_dict}\n")

    model_checkpoint_callback = ModelCheckpoint(filepath='stacn_best_model.keras', save_best_only=True, monitor='val_accuracy', mode='max', verbose=1)
    early_stopping_callback = EarlyStopping(monitor='val_loss', patience=15, verbose=1, restore_best_weights=True)

    print("--- Starting Model Training (with class weights) ---")
    history = stacn_model.fit(X_train, y_train,
                              epochs=CONFIG['epochs'],
                              batch_size=CONFIG['batch_size'],
                              validation_data=(X_test, y_test),
                              verbose=1,
                              callbacks=[model_checkpoint_callback, early_stopping_callback],
                              class_weight=class_weight_dict
                             )

    evaluate_and_visualize_model(stacn_model, history, X_test, y_test, emotion_labels, CONFIG)


if __name__ == '__main__':
    main()
```
